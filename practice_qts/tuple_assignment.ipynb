```python
# Assignment 15 :
def min_in_tpl(tpl):
    return min(tpl)

def max_in_tpl(tpl):
    return max(tpl)

def sum_in_tpl(tpl):
    return sum(tpl)

sample_tpl = (1,2,3,4,5)
print(f"Minimum :{min_in_tpl(sample_tpl)}")
print(f"Maximum :{max_in_tpl(sample_tpl)}")
print(f"SUM :{sum_in_tpl(sample_tpl)}")
```
